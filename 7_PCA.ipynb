1.Maggie is a real-estate agent and she is puzzled about why some of the properties her company manages have not been sold for more than six months. Help Maggie in providing the insights why the properties are not getting sold.

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# 1. Create sample real-estate dataset
data = {
    "Area": [1200, 1500, 1800, 900, 1100, 2000, 1600, 1300, 2200, 1000,
             1700, 1450, 1900, 1250, 2100],

    "Bedrooms": [2, 3, 3, 2, 2, 4, 3, 2, 4, 2,
                 3, 3, 4, 2, 4],

    "Bathrooms": [2, 2, 3, 1, 2, 3, 2, 2, 3, 1,
                  2, 2, 3, 2, 3],

    "Age": [5, 8, 3, 20, 15, 2, 7, 12, 4, 25,
            6, 10, 3, 18, 5],

    "Distance": [5, 8, 4, 15, 12, 3, 7, 10, 2, 18,
                 6, 9, 4, 14, 3],

    "CrimeRate": [2, 3, 2, 7, 6, 1, 3, 5, 1, 8,
                  2, 4, 2, 7, 1],

    "Price": [70, 85, 110, 45, 60, 130, 90, 65, 150, 50,
              100, 80, 120, 55, 140],

    "Days_On_Market": [120, 210, 90, 250, 230, 60, 180, 200, 45, 270,
                       100, 190, 80, 240, 50]
}
df = pd.DataFrame(data)

print("Real Estate Dataset:")
print(df)


# 2. Identify properties unsold for more than 6 months
df["Unsold_6_Months"] = df["Days_On_Market"] > 180

print("\nProperties unsold for more than 6 months:")
print(df[df["Unsold_6_Months"]])


# 3. Select property characteristics for PCA
features = [
    "Area",
    "Bedrooms",
    "Bathrooms",
    "Age",
    "Distance",
    "CrimeRate",
    "Price"
]
X = df[features]


# 4. Standardize the data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)


# 5. Apply PCA
pca = PCA()
X_pca = pca.fit_transform(X_scaled)


# 6. Display explained variance
print("\nExplained Variance:")
for i, value in enumerate(pca.explained_variance_ratio_):
    print(f"PC{i+1}: {value:.2%}")

    
# 7. Display PCA loadings
loadings = pd.DataFrame(
    pca.components_.T,
    columns=[f"PC{i+1}" for i in range(len(features))],
    index=features
)
print("\nPCA Loadings:")
print(loadings)


# 8. Identify the most important variables
print("\nMost Important Variables:")
for pc in ["PC1", "PC2", "PC3"]:
    variable = loadings[pc].abs().idxmax()
    loading = loadings.loc[variable, pc]

    print(f"{pc} -> {variable} (Loading = {loading:.3f})")


# 9. Compare PCA scores of properties
pca_scores = pd.DataFrame(
    X_pca,
    columns=[f"PC{i+1}" for i in range(len(features))]
)
pca_scores["Unsold_6_Months"] = df["Unsold_6_Months"]

print("\nAverage PCA Scores:")
print(
    pca_scores.groupby("Unsold_6_Months")[["PC1", "PC2", "PC3"]].mean()
)
```

2. For the given dataset, identify which variable is pricipal component.

```python
import pandas as pd

# PCA Loading Matrix
data = {
    "Variable": [
        "Climate",
        "Housing",
        "Health",
        "Crime",
        "Transportation",
        "Education",
        "Arts",
        "Recreation",
        "Economy"
    ],
    "PC1": [
        0.190, 0.544, 0.782, 0.365,
        0.585, 0.394, 0.985, 0.520, 0.142
    ],
    "PC2": [
        0.017, 0.020, -0.605, 0.294,
        0.085, -0.273, 0.126, 0.402, 0.150
    ],
    "PC3": [
        0.207, 0.204, 0.144, 0.585,
        0.234, 0.027, -0.111, 0.519, 0.239
    ]
}

df = pd.DataFrame(data)

print("\nPCA LOADING MATRIX:")
print(df.to_string(index=False))


# Find the most important variable for each Principal Component
print("\nMOST IMPORTANT VARIABLES:")
for pc in ["PC1", "PC2", "PC3"]:
    index = df[pc].abs().idxmax()
    variable = df.loc[index, "Variable"]
    loading = df.loc[index, pc]
    print(
        f"{pc} -> {variable} "
        f"(Loading = {loading:.3f})"
    )
```
